```python
%%writefile app.py
import streamlit as st
import pandas as pd
import os
import time

# Set page config
st.set_page_config(
    page_title="Edge-IDPS Security Dashboard",
    page_icon="🛡️",
    layout="wide"
)

# Header
st.title("🛡️ Edge-IDPS Real-Time Threat Monitoring Center")
st.markdown("Automated Intrusion Detection, Fuzzy Risk Assessment, and Prevention Engine")

# Path to the live log generated by Member 3
LOG_FILE_PATH = "/content/drive/MyDrive/Edge-IDPS-Data/processed_data/live_stream_output.csv"

@st.cache_data(ttl=1)
def load_data():
    if os.path.exists(LOG_FILE_PATH):
        try:
            df = pd.read_csv(LOG_FILE_PATH)
            return df
        except Exception:
            return pd.DataFrame()
    return pd.DataFrame()

df = load_data()

if df.empty:
    st.warning("⏳ Waiting for streaming data from Member 3's pipeline...")
    st.info("Make sure Member 3's streaming script is running and logging to Google Drive.")
else:
    # Top KPI Metrics Row
    total_packets = len(df)
    critical_threats = len(df[df['risk_level'] == 'CRITICAL'])
    high_threats = len(df[df['risk_level'] == 'HIGH'])
    allowed_traffic = len(df[df['action'] == 'ALLOW'])

    col1, col2, col3, col4 = st.columns(4)
    col1.metric("Total Packets Inspected", total_packets)
    col2.metric("Critical Threats (Blocked)", critical_threats, delta_color="inverse")
    col3.metric("High/Medium Risks", high_threats)
    col4.metric("Normal Traffic Allowed", allowed_traffic)

    st.markdown("---")

    # Layout: Two Columns for Charts & Tables
    left_col, right_col = st.columns([1, 1])

    with left_col:
        st.subheader("🚨 Risk Level Distribution")
        risk_counts = df['risk_level'].value_counts()
        st.bar_chart(risk_counts)

        st.subheader("⚡ Automated Mitigation Actions")
        action_counts = df['action'].value_counts()
        st.bar_chart(action_counts)

    with right_col:
        st.subheader("📈 Top Detected Threat Classes")
        class_counts = df['predicted_class'].value_counts().head(5)
        st.dataframe(class_counts, use_container_width=True)

        st.subheader("🔥 Average Risk Score by Attack Vector")
        if 'risk_score' in df.columns:
            avg_risk = df.groupby('predicted_class')['risk_score'].mean().reset_index()
            st.dataframe(avg_risk.sort_values(by='risk_score', ascending=False), use_container_width=True)

    st.markdown("---")

    # Live Log Table
    st.subheader("📋 Live Real-Time Packet Stream Log")

    # Highlight critical rows visually
    def highlight_severity(val):
        color = ''
        if val == 'CRITICAL':
            color = 'background-color: #ff4b4b; color: white;'
        elif val == 'HIGH':
            color = 'background-color: #ffa500; color: black;'
        elif val == 'MEDIUM':
            color = 'background-color: #ffff00; color: black;'
        elif val == 'LOW':
            color = 'background-color: #00e676; color: black;'
        return color

    st.dataframe(
        df.tail(15).iloc[::-1], # Show newest 15 packets first
        use_container_width=True
    )

# Auto-refresh dashboard every 2 seconds
time.sleep(2)
st.rerun()
```
