```python
# 1. Count how many times each class appears
class_counts = y.value_counts()

# 2. Identify classes that appear fewer than 2 times (or fewer than 5)
rare_classes = class_counts[class_counts < 2].index

# 3. Filter out those single-sample rows from both X and y
valid_indices = ~y.isin(rare_classes)
X_clean = X[valid_indices]
y_clean = y[valid_indices]

# 4. Now run train_test_split on the cleaned data
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42, stratify=y_clean
)

print("Split completed successfully!")
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

# 1. Get unique classes and their counts in the NumPy array y
classes, counts = np.unique(y, return_counts=True)

# 2. Identify classes that appear fewer than 2 times
rare_classes = classes[counts < 2]

# 3. Create a boolean mask to filter out single-sample classes from X and y
mask = ~np.isin(y, rare_classes)
X_clean = X[mask]
y_clean = y[mask]

# 4. Now run train_test_split on the cleaned arrays
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42, stratify=y_clean
)

print("Split completed successfully!")
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```
